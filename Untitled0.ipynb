```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Helper function to dynamically locate image files in Colab or local directory
def get_image_path(filename):
    colab_path = os.path.join('/content', filename)
    if os.path.exists(colab_path):
        return colab_path
    elif os.path.exists(filename):
        return filename
    else:
        raise FileNotFoundError(f"Please upload '{filename}' to your working directory or Colab environment.")
```

## Exercise 1: Color Order Swapping (BGR to RGB)

OpenCV loads images in BGR format by default. To display them correctly with matplotlib, we must reverse the color channels to RGB.

```python
# Find and load the main image
image_path = get_image_path('image.png')
img_bgr = cv2.imread(image_path)

# Swap channels (BGR to RGB) using numpy slicing
img_rgb = img_bgr[:, :, ::-1]

# Display BGR vs RGB side-by-side
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(img_bgr)
axes[0].set_title("Original Loaded BGR (Incorrect Colors)")
axes[0].axis('off')

axes[1].imshow(img_rgb)
axes[1].set_title("Swapped Channels to RGB (Correct Colors)")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## Exercise 2: Separating Color Channels

We will extract and display the Red, Green, and Blue channels of the image independently as grayscale representations.

```python
# Extract channels from the RGB image
R_channel = img_rgb[:, :, 0]
G_channel = img_rgb[:, :, 1]
B_channel = img_rgb[:, :, 2]

# Display channels independently in grayscale
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(R_channel, cmap='gray')
axes[0].set_title("Red Channel")
axes[0].axis('off')

axes[1].imshow(G_channel, cmap='gray')
axes[1].set_title("Green Channel")
axes[1].axis('off')

axes[2].imshow(B_channel, cmap='gray')
axes[2].set_title("Blue Channel")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

## Exercise 3: ROI Mask Generation and Application

We define a $512 \times 512$ binary mask representing a Region of Interest (ROI) between rows 200 to 400 and columns 200 to 350, then apply it to each channel of our resized image.

```python
# Resize image to 512x512
im_resized = cv2.resize(img_rgb, (512, 512))

# a) Define 512x512 mask
mask = np.zeros((512, 512), dtype=np.float32)
mask[200:401, 200:351] = 1.0

# c) Apply the mask to each channel of the color image
masked_im = np.zeros_like(im_resized)
for i in range(3):
    masked_im[:, :, i] = im_resized[:, :, i] * mask

# b & d) Display the mask, the original image, and the masked result
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(mask, cmap='gray')
axes[0].set_title("a, b) 512x512 Binary Mask")
axes[0].axis('off')

axes[1].imshow(im_resized)
axes[1].set_title("Original Image (Resized)")
axes[1].axis('off')

axes[2].imshow(masked_im.astype(np.uint8))
axes[2].set_title("c, d) Masked Output Image")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

## Exercise 4: Image Blending / Fusion

Here we load two different images (`image.jpg` and `image (1).jpg`), normalize them to the floating-point range `[0.0, 1.0]`, and calculate their pixel-wise mean to create a smooth blend.

```python
# Load two separate source images
img1_path = get_image_path('image.jpg')
img2_path = get_image_path('image (1).jpg')

face1_bgr = cv2.imread(img1_path)
face2_bgr = cv2.imread(img2_path)

# Convert both to RGB
face1_rgb = cv2.cvtColor(face1_bgr, cv2.COLOR_BGR2RGB)
face2_rgb = cv2.cvtColor(face2_bgr, cv2.COLOR_BGR2RGB)

# Ensure both images are of the same size before blending
h, w, _ = face1_rgb.shape
face2_rgb_resized = cv2.resize(face2_rgb, (w, h))

# Normalize images into float representation in range [0.0, 1.0]
face1_float = face1_rgb.astype(np.float32) / 255.0
face2_float = face2_rgb_resized.astype(np.float32) / 255.0

# Blend them by computing their mean
blended_face = (face1_float + face2_float) / 2.0

# Display the fusion results
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(face1_float)
axes[0].set_title("Face Image 1")
axes[0].axis('off')

axes[1].imshow(face2_float)
axes[1].set_title("Face Image 2")
axes[1].axis('off')

axes[2].imshow(blended_face)
axes[2].set_title("Fused (Mean Blend) Output")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
